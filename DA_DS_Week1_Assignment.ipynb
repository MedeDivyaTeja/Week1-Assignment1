###**Slicing**
####8. Explain what slicing is. How does slicing work in:  


*   Python lists
*   NumPy arrays

**Answer:**
Slicing is used to access a part of a sequence like a list or array.

Python List Slicing:








```python
numbers = [1,2,3,4,5]

print(numbers[1:4])
```

NumPy Array Slicing:

```python
import numpy as np

arr = np.array([10,20,30,40,50])

print(arr[1:4])
```
