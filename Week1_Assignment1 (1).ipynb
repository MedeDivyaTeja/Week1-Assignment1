###**11. Sign Checker:**

####Write a program that takes a number from the user, checks whether the number is positive, negative, or zero, and displays an appropriate message.

```python
# Program to check whether a number is positive, negative, or zero

num = int(input("Enter a number: "))

if num > 0:
    print("Positive Number")

elif num < 0:
    print("Negative Number")

else:
    print("Zero")
```

###**12. Stats Function:**

####Write a program that takes a number from the user, checks whether the number is positive, negative, or zero, and displays an appropriate message.

```python
# Function to calculate sum and average

def stats(numbers):

    total = sum(numbers)
    average = total / len(numbers)

    return total, average


nums = [10, 20, 30, 40]

result = stats(nums)

print("Sum:", result[0])
print("Average:", result[1])
```

###**13. Even Numbers:**

####Write a program using a for loop to print the first n even numbers, where n is taken as input from the user.

```python
# Program to print first n even numbers

n = int(input("Enter value of n: "))

for i in range(1, n + 1):

    print(i * 2)
```

###**14. Manual Palindrome:**

####Write a program to check whether a given string is a palindrome without using any built-in reverse function.

```python
# Program to check palindrome without using reverse function

text = input("Enter a string: ")

reverse = ""

for char in text:

    reverse = char + reverse


if text == reverse:
    print("Palindrome")

else:
    print("Not Palindrome")
```
